```python
import subprocess, sys
from pathlib import Path

# Install hysprint_utils from shared/ relative to this notebook, then add it
# to sys.path directly. A fresh editable install's import hook is only
# processed by site.py at interpreter startup, so pip install alone would
# require restarting the kernel before hysprint_utils becomes importable;
# inserting it into sys.path here makes the current kernel see it immediately.
_shared = (Path.cwd() / "../../shared").resolve()
subprocess.run(
    [sys.executable, "-m", "pip", "install", "-q", str(_shared)],
    check=True,
)
if str(_shared) not in sys.path:
    sys.path.insert(0, str(_shared))
del _shared

from app import setup_app
from IPython.display import HTML, display

from hysprint_utils.access_token import log_notebook_usage

log_notebook_usage()
```
